Used to calculate the maximum amount of time it would take direct sound to reach the microphones, windowing out any soundreaching the mic after this (believed to be early/late reflections)

First Method - Using Room Radius

```python
import numpy as np

T20 = 0.17
L = 5.305
W = 4.909
H = 2.343

V = (L*W*H)
print(V)

S = (2*(L*W) + 2*(L*H) + 2*(W*H))

A = (0.16 * V)/T20

alpha = A/S

Q = 1

c = 343

fs = 48000



Room_const = (S*alpha)/(1-alpha)
print(Room_const)

Room_radius = np.sqrt((Room_const * Q)/(16*np.pi))
print(Room_radius)

time_2_first_reflection = Room_radius/c
print(time_2_first_reflection)

n_samples = fs*time_2_first_reflection
print(int(n_samples), "From onset")
```

Second Method - Using room dimensions to determine first possible reflection length
Image source method

```python
source_pos = np.array([1.0, W/2, 1.5])
mic_pos = np.array([1.0 + 1.7, W/2, 1.5])
room_dims = np.array([L, W, H])

direct_dist = np.linalg.norm(source_pos - mic_pos)

distances = []
for axis in range(3):
    for direction in [0, room_dims[axis]]:
        mirrored = source_pos.copy()
        mirrored[axis] = 2 * direction - mirrored[axis]
        distances.append(np.linalg.norm(mirrored - mic_pos))
        
first_reflection_dist = min(distances)
first_reflection_time = first_reflection_dist / c
direct_time = direct_dist / c

window_length_samples = int((first_reflection_time - direct_time) * fs)

print(f"Direct sound arrives at: {direct_time*1000:.2f}ms")
print(f"First reflection at: {first_reflection_time*1000:.2f}ms")
print(f"Window length: {window_length_samples} samples ({window_length_samples/fs*1000:.2f}ms)")
```
